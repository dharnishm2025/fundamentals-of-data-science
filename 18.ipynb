# Experiment 18: KNN Algorithm

Executed notebook with output.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,classification_report
d=load_iris(); Xtr,Xte,ytr,yte=train_test_split(d.data,d.target,test_size=.25,random_state=42,stratify=d.target); s=StandardScaler(); Xtr=s.fit_transform(Xtr); Xte=s.transform(Xte); model=KNeighborsClassifier(n_neighbors=5).fit(Xtr,ytr); pred=model.predict(Xte); print(f'Accuracy: {accuracy_score(yte,pred):.2%}'); print(classification_report(yte,pred,target_names=d.target_names))
```
