# Experiment 16: EDA - Quantitative and Qualitative Analysis

Executed notebook with output.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
rng=np.random.default_rng(5); df=pd.DataFrame({'Department':rng.choice(['CSE','IT','ECE'],80),'Study_Hours':rng.normal(4,1.2,80).clip(1,8),'Marks':rng.normal(72,10,80).clip(35,98)}); print('Quantitative summary:'); display(df[['Study_Hours','Marks']].describe()); print('Qualitative frequency:'); print(df.Department.value_counts()); fig,ax=plt.subplots(1,2,figsize=(11,4)); sns.histplot(df.Marks,kde=True,ax=ax[0]); sns.boxplot(data=df,x='Department',y='Marks',ax=ax[1]); plt.tight_layout(); plt.show()
```
