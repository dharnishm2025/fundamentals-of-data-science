# Experiment 19: Logistic Regression

Executed notebook with output.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
d=load_breast_cancer(); Xtr,Xte,ytr,yte=train_test_split(d.data,d.target,test_size=.25,random_state=42,stratify=d.target); s=StandardScaler(); Xtr=s.fit_transform(Xtr); Xte=s.transform(Xte); model=LogisticRegression(max_iter=2000,random_state=42).fit(Xtr,ytr); pred=model.predict(Xte); print(f'Accuracy: {accuracy_score(yte,pred):.2%}'); print('Confusion matrix:\n',confusion_matrix(yte,pred))
```
