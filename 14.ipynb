# Experiment 14: Feature Scaling

Executed notebook with output.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler,StandardScaler
df=pd.DataFrame({'Age':[18,22,30,45,60],'Income':[18000,25000,42000,65000,90000]}); print('Original:'); display(df); print('Min-Max scaled:'); display(pd.DataFrame(MinMaxScaler().fit_transform(df),columns=df.columns)); print('Standardized:'); display(pd.DataFrame(StandardScaler().fit_transform(df),columns=df.columns))
```
