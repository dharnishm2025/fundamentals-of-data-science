# Experiment 17: Linear Regression

Executed notebook with output.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.linear_model import LinearRegression
hours=np.array([1,2,3,4,5,6,7,8]).reshape(-1,1); marks=np.array([42,48,55,61,67,74,82,88]); model=LinearRegression().fit(hours,marks); pred=model.predict(hours); print(f'Equation: Marks = {model.coef_[0]:.2f} x Hours + {model.intercept_:.2f}'); print(f'Prediction for 6.5 hours: {model.predict([[6.5]])[0]:.2f}'); plt.scatter(hours,marks,label='Actual'); plt.plot(hours,pred,c='red',label='Regression line'); plt.xlabel('Study hours'); plt.ylabel('Marks'); plt.legend(); plt.show()
```
