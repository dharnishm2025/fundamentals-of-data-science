# Experiment 13: Handling Missing and Inappropriate Data

Executed notebook with output.

```python
import numpy as np
import pandas as pd
df=pd.DataFrame({'Name':['Asha','Bala','Chitra','Dev'],'Age':[20,np.nan,-5,22],'Marks':[82,75,np.nan,91]}); print('Original:'); display(df); df.loc[df.Age<0,'Age']=np.nan; df['Age']=df.Age.fillna(df.Age.median()); df['Marks']=df.Marks.fillna(df.Marks.mean()); print('Cleaned:'); display(df); print(df.isna().sum())
```
