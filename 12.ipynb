# Experiment 12: Outlier Detection

Executed notebook with output.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
df=pd.DataFrame({'Marks':[45,52,48,55,51,49,54,50,53,98]}); q1,q3=df.Marks.quantile([.25,.75]); iqr=q3-q1; lo,hi=q1-1.5*iqr,q3+1.5*iqr; df['Outlier']=(df.Marks<lo)|(df.Marks>hi); print(f'IQR bounds: {lo:.2f} to {hi:.2f}'); display(df); plt.boxplot(df.Marks,vert=False); plt.title('Outlier Detection using IQR'); plt.show()
```
