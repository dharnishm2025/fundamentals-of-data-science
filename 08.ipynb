# Experiment 8: Z Test

Executed notebook with output.

```python
import numpy as np
from scipy.stats import norm
sample=np.array([52,49,54,51,53,50,55,52,48,54]); mu=50; sigma=4
z=(sample.mean()-mu)/(sigma/np.sqrt(len(sample))); p=2*(1-norm.cdf(abs(z)))
print(f'Sample mean: {sample.mean():.2f}'); print(f'Z statistic: {z:.4f}'); print(f'P value: {p:.4f}'); print('Decision at alpha=0.05:', 'Reject H0' if p<.05 else 'Fail to reject H0')
```
