# Experiment 20: K-Means Clustering

Executed notebook with output.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
X,_=make_blobs(n_samples=150,centers=3,cluster_std=.85,random_state=42); model=KMeans(n_clusters=3,n_init=10,random_state=42).fit(X); print('Cluster centers:\n',np.round(model.cluster_centers_,2)); plt.scatter(X[:,0],X[:,1],c=model.labels_,cmap='viridis',s=35); plt.scatter(model.cluster_centers_[:,0],model.cluster_centers_[:,1],c='red',marker='X',s=180,label='Centroids'); plt.title('K-Means Clustering'); plt.legend(); plt.show()
```
