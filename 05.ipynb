Executed notebook with output.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
rng=np.random.default_rng(42)
df=pd.DataFrame({'Pregnancies':rng.integers(0,10,30),'Glucose':rng.integers(70,190,30),'BMI':np.round(rng.uniform(18,45,30),1),'Age':rng.integers(21,70,30)})
df['Outcome']=((df.Glucose>140)&(df.BMI>28)).astype(int)
print('First five rows:'); display(df.head())
print('Dataset information:'); df.info()
print('Summary statistics:'); display(df.describe())
print('Missing values:'); print(df.isna().sum())
df.hist(figsize=(9,6),edgecolor='black'); plt.suptitle('Diabetes Dataset - Initial Exploration'); plt.tight_layout(); plt.show()
```
