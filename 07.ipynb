# Experiment 7: Random Sampling and Sampling Distribution

Executed notebook with output.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
rng=np.random.default_rng(21); population=rng.normal(70,12,10000); sample=rng.choice(population,50,replace=False); means=[rng.choice(population,50,replace=False).mean() for _ in range(1000)]
print(f'Population mean: {population.mean():.2f}'); print(f'Sample mean: {sample.mean():.2f}'); print(f'Mean of sampling distribution: {np.mean(means):.2f}')
fig,ax=plt.subplots(1,2,figsize=(11,4)); ax[0].hist(population,bins=35); ax[0].set_title('Population'); ax[1].hist(means,bins=30); ax[1].set_title('Sampling Distribution of Means'); plt.show()
```
