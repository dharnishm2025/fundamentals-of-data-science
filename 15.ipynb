# Experiment 15: Data Preprocessing

Executed notebook with output.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
df=pd.DataFrame({'City':['Chennai','Madurai','Chennai','Coimbatore'],'Age':[20,np.nan,25,30],'Income':[30000,45000,np.nan,60000]}); print('Raw data:'); display(df); df['Age']=df.Age.fillna(df.Age.median()); df['Income']=df.Income.fillna(df.Income.mean()); clean=pd.get_dummies(df,columns=['City'],dtype=int); clean[['Age','Income']]=StandardScaler().fit_transform(clean[['Age','Income']]); print('Preprocessed data:'); display(clean)
```
